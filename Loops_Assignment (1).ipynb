## School Marks Report
### Goal:

    Take input for multiple students’ marks in 3 subjects and calculate their total and average.

### Steps:

    Outer loop → for each student (3 students).
    
    Inner loop → take marks for 3 subjects.
    
    Store each student’s total in a list.
    
    Print all totals and class average.

### Example:
    Enter marks for Student 1:
    Subject 1: 80
    Subject 2: 75
    Subject 3: 90
    Enter marks for Student 2:
    Subject 1: 60
    Subject 2: 70
    Subject 3: 65
    → Output: [245, 195]

```python
total = []
class_total = 0

for i in range(1, 4):
    print(f"Enter marks for Student {i}:")
    student_total = 0

    for j in range(1, 4):
        sub_marks = int(input(f"Subject {j}: "))
        student_total += sub_marks

    total.append(student_total)
    class_total += student_total

average = class_total / 3

print("Student totals:", total)
print("Class average:", average)
        
```

## Shop Sales Tracker
### Goal:

    Track daily sales of 3 products over 5 days and calculate the total revenue.

### Steps:

    Outer loop → iterate over 5 days.
    
    Inner loop → take sales amount for 3 products each day.
    
    Store daily total in a list.
    
    If any product sale = 0 → continue (skip that product).
    
    Print all daily totals and grand total.

### Example:
    Day 1:
    Product 1: 200
    Product 2: 150
    Product 3: 0 → skipped
    Day 2:
    Product 1: 300
    Product 2: 250
    Product 3: 400
    → Daily Totals = [350, 950], Grand Total = 1300

```python
total = []
daily_total = []

for i in range(1, 6):
    print(f"Day {i}")
    day_total = 0

    for j in range(1, 4):
        sale_amount = int(input(f"Product {j}: "))

        if sale_amount <= 0:
            print("Skipped")
            continue
        else:
            total.append(sale_amount)
            day_total += sale_amount

    daily_total.append(day_total)

grand_total = sum(total)

print("Daily Totals =", daily_total)
print("Grand Total =", grand_total)

    
    
    
```

## Holiday Package Evaluator
### Goal:

Compare multiple travel packages across features like hotel, transport, food.

### Steps:

Outer loop → 3 packages.

Inner loop → 3 features.

Input rating out of 5.

Skip if rating = 0.

Print average rating per package.

### Example:
Package 1: [4, 5, 5] → Avg = 4.67  
Package 2: [3, 4, 0] → Avg = 3.5

```python

for i in range(1, 4):
    ratings = []
    print(f"\nPackage {i}")

    for j in range(1, 4):
        rating = int(input(f"Enter rating for feature {j} (0-5): "))

        if rating == 0:
            continue

        ratings.append(rating)

    if len(ratings) > 0:
        avg = sum(ratings) / len(ratings)
        print("Ratings:", ratings)
        print("Average Rating:", round(avg, 2))
    else:
        print("No ratings available")

            
```

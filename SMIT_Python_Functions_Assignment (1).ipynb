## Question 28 — Smart Shopping Billing System

**Problem:**  
Build a shopping billing system that can handle multiple products.

**Requirements:**
- Allow the user (or hard-code) multiple products.
- Calculate:
  - Individual product totals
  - Overall subtotal
  - Discount
  - Tax
  - Final bill
- Use **multiple functions** to separate responsibilities.
- At least one function must use `*args` or a flexible-argument design.
- Display a professional-looking bill.

**Focus:** Real-world application + multiple functions + logic building

````python
# Write your code below

def product_info(name, price, quantity):
    total = price * quantity
    return name, price, quantity, total


def calculate_total(*args):
    return sum(args)


def calculate_discount(subtotal):
    if subtotal >= 5000:
        return subtotal * 0.10
    else:
        return 0


def calculate_tax(amount):
    return amount * 0.05


def final_bill(subtotal, discount, tax):
    return subtotal - discount + tax


def display_bill(products):
    print("\n========== SHOPPING BILL ==========")

    totals = []

    for product in products:
        name, price, quantity, total = product

        print(f"\nProduct: {name}")
        print(f"Price: {price}")
        print(f"Quantity: {quantity}")
        print(f"Total: {total}")

        totals.append(total)

    subtotal = calculate_total(*totals)
    discount = calculate_discount(subtotal)
    amount_after_discount = subtotal - discount
    tax = calculate_tax(amount_after_discount)

    bill = final_bill(subtotal, discount, tax)

    print("\n===================================")
    print("Subtotal:", subtotal)
    print("Discount:", discount)
    print("Tax:", tax)
    print("-----------------------------------")
    print("Final Bill:", bill)
    print("===================================")



products = [
    product_info("Keyboard", 2000, 2),
    product_info("Mouse", 1000, 1),
    product_info("Headphones", 3000, 1)
]


display_bill(products)
````

## Question 29 — Student Result Management System

**Problem:**  
Create a complete student result system using functions.

**The system should:**
1. Accept student information
2. Accept multiple subject marks
3. Calculate total
4. Calculate percentage
5. Determine grade
6. Determine pass/fail
7. Display a final result card

**Requirements:**
- Use **multiple functions**
- Use arguments and `return`
- Use lists
- Use conditions and loops
- Use at least one advanced concept (`*args`, `**kwargs`, nested functions, or lambda)
- Do **not** write the entire program inside one function

**Focus:** Combining multiple function concepts

````python
# Write your code below

def student_info(name, roll_no, department):
    return name, roll_no, department


def accept_marks(*marks):
    return list(marks)


def calculate_total(marks):
    return sum(marks)


def calculate_percentage(total, subjects):
    return total / subjects


def calculate_grade(percentage):
    if percentage >= 90:
        return "A+"
    elif percentage >= 80:
        return "A"
    elif percentage >= 70:
        return "B"
    elif percentage >= 60:
        return "C"
    elif percentage >= 50:
        return "D"
    else:
        return "F"


def check_result(marks):
    for mark in marks:
        if mark < 40:
            return "Fail"
    return "Pass"


def display_result(student, marks):
    name, roll_no, department = student

    total = calculate_total(marks)
    percentage = calculate_percentage(total, len(marks))
    grade = calculate_grade(percentage)
    result = check_result(marks)

    print("\n========== STUDENT RESULT CARD ==========")
    print("Name:", name)
    print("Roll Number:", roll_no)
    print("Department:", department)

    print("\n---------- Subject Marks ----------")

    for i, mark in enumerate(marks, start=1):
        print(f"Subject {i}: {mark}")

    print("-----------------------------------")
    print("Total Marks:", total, "/", len(marks) * 100)
    print("Percentage:", percentage, "%")
    print("Grade:", grade)
    print("Result:", result)
    print("===================================")


student = student_info("Tayyab", 101, "IT")


marks = accept_marks(85, 90, 78, 88, 92)


display_result(student, marks)

````

````python
# Write your code below

def customer_info(name, phone, address="Karachi"):
    return {
        "name": name,
        "phone": phone,
        "address": address
    }


def product_info():
    products = {
        1: {"name": "Keyboard", "price": 2000},
        2: {"name": "Mouse", "price": 1000},
        3: {"name": "Headphones", "price": 3000},
        4: {"name": "Monitor", "price": 15000}
    }

    return products


def calculate_total(price, quantity):
    return price * quantity


def calculate_subtotal(*args):
    return sum(args)


def calculate_discount(subtotal):
    if subtotal >= 10000:
        return subtotal * 0.10
    elif subtotal >= 5000:
        return subtotal * 0.05
    else:
        return 0


def calculate_tax(amount, rate=0.05):
    return amount * rate


def delivery_charges(subtotal):
    if subtotal >= 10000:
        return 0
    else:
        return 300


def final_bill(subtotal, discount, tax, delivery):
    return subtotal - discount + tax + delivery


def place_order(**kwargs):
    return kwargs


def display_summary(customer, order, products):

    
    def print_line():
        print("---------------------------------------")

    subtotal = calculate_subtotal(
        *[item["total"] for item in order]
    )

    discount = calculate_discount(subtotal)

    after_discount = subtotal - discount

    tax = calculate_tax(after_discount)

    delivery = delivery_charges(subtotal)

    total = final_bill(
        subtotal, discount, tax, delivery
    )

    print("\n========== E-COMMERCE INVOICE ==========")

    print("Customer:", customer["name"])
    print("Phone:", customer["phone"])
    print("Address:", customer["address"])

    print_line()

    print("ORDER DETAILS")

    for item in order:
        print(
            item["name"],
            "| Price:", item["price"],
            "| Quantity:", item["quantity"],
            "| Total:", item["total"]
        )

    print_line()

    print("Subtotal:", subtotal)
    print("Discount:", discount)
    print("Tax:", tax)
    print("Delivery Charges:", delivery)

    print_line()

    print("Final Bill:", total)
    print("=======================================")


# Main program

customer = customer_info(
    "Tayyab",
    "03001234567"
)

products = product_info()

order = []

# Lambda function
get_price = lambda price, quantity: price * quantity

print("AVAILABLE PRODUCTS")

for key, value in products.items():
    print(key, value["name"], "-", value["price"])


while True:

    choice = int(input("\nEnter product number (0 to finish): "))

    if choice == 0:
        break

    if choice not in products:
        print("Invalid product selection!")
        continue

    quantity = int(input("Enter quantity: "))

    if quantity <= 0:
        print("Quantity must be greater than zero!")
        continue

    name = products[choice]["name"]
    price = products[choice]["price"]

    total = get_price(price, quantity)

    order.append({
        "name": name,
        "price": price,
        "quantity": quantity,
        "total": total
    })


if len(order) > 0:

    order_details = place_order(
        customer=customer["name"],
        total_products=len(order)
    )

    display_summary(customer, order, products)

else:
    print("No products selected. Order cancelled.")
````
